# Split-Half Reliability Noise Ceiling Across ROIs

## Functions defined

```python
def split_half_reliability_ROI(trials_df, roi_names, subject_masks):
    # Initialize lists to store even and odd split data
    even = []
    odd = []
    
    # Split trials based on identifier
    for video, group in trials_df.groupby('identifier'):
        group = group.reset_index(drop=True)
        if len(group) > 1:
            odd_rows = group[group.index % 2 != 0]
            even_rows = group[group.index % 2 == 0]
            odd.append(odd_rows['betas'].mean(axis=0).flatten())  # Voxel-wise average for odd trials
            even.append(even_rows['betas'].mean(axis=0).flatten())  # Voxel-wise average for even trials

    # Convert even and odd lists to arrays
    even = np.array(even).T
    odd = np.array(odd).T # Shape: (voxels, trials)

    # Initialize dictionary to store split-half Pearson correlation for each ROI
    split_half_r = {}

    for roi_name in roi_names:
        # Retrieve voxel mask for the current ROI and flatten to get relevant voxel indices
        mask_data = subject_masks[roi_name[0]][roi_name[1]].get_fdata().flatten()
        roi_voxel_indices = np.where(mask_data > 0)[0]  # Indices where mask is non-zero
        
        # Calculate split-half Pearson correlation for each voxel in the ROI
        r_values = []
        for voxel_id in roi_voxel_indices: #vmpfc voxels sometimes give warning..
            #print(even[voxel_id, :])
            if not np.isfinite(even[voxel_id, :]).all() or not np.isfinite(odd[voxel_id, :]).all():
                continue
            if np.all(even[voxel_id, :] == even[voxel_id, 0]) or np.all(odd[voxel_id, :] == odd[voxel_id, 0]):
                continue
            r, _ = scipy.stats.pearsonr(even[voxel_id, :], odd[voxel_id, :])
            #r = vectorized_metric(even[voxel_id, :], odd[voxel_id, :], "pearsonr")
            r_values.append(r)
        r_values = np.array(r_values)
        r_values[r_values < 0] = 0

        # Average r values across all voxels within the ROI
        split_half_r[roi_name[1]] = np.nanmean(r_values) if len(r_values) else np.nan

    return split_half_r
```

# Split-Half Reliability Whole-brain (grouped) plot

## Function Defined

```python
from general_utils import plot_on_surf

def split_half_reliability_wholebrain(trials_df):
    """
    Computes split-half reliability (Pearson r) for each voxel over all trials.
    Assumes trials_df has a column 'identifier' (for each video) and a column 'betas'
    that contains voxel-wise beta estimates (as arrays) for that trial.
    
    Returns:
        reliability: 1D NumPy array of reliability values, one per voxel.
    """
    even = []
    odd = []
    
    # Split trials into even and odd for each video (identifier)
    for video, group in trials_df.groupby('identifier'):
        group = group.reset_index(drop=True)
        if len(group) > 1:
            odd_rows = group[group.index % 2 != 0]
            even_rows = group[group.index % 2 == 0]
            # Average the beta estimates across odd and even trials, respectively. (note that makes odd 2 rows, and even 3)
            odd.append(odd_rows['betas'].mean(axis=0).flatten())
            even.append(even_rows['betas'].mean(axis=0).flatten())
    
    # Convert lists to arrays; shape will be (voxels, number_of_splits)
    even = np.array(even).T
    odd  = np.array(odd).T
    
    n_voxels = even.shape[0]
    reliability = np.empty(n_voxels)
    reliability.fill(np.nan)
    
    p_value = np.empty(n_voxels)
    p_value.fill(np.nan)

    # Compute voxelwise Pearson correlation between even and odd splits.
    for voxel_id in range(n_voxels):
        # Skip voxels with non-finite data or with no variance.
        if (not np.all(np.isfinite(even[voxel_id, :])) or 
            not np.all(np.isfinite(odd[voxel_id, :])) or
            np.all(even[voxel_id, :] == even[voxel_id, 0]) or 
            np.all(odd[voxel_id, :] == odd[voxel_id, 0])):
            continue
        r, p = scipy.stats.pearsonr(even[voxel_id, :], odd[voxel_id, :])
        reliability[voxel_id] = r
        p_value[voxel_id] = p

    return reliability, p_value
```

# Split-Half RSA Noise Ceiling Across ROIs

## Functions Defined

```python
def split_half_RSA_ROI(trials_df, roi_names, subject_masks, affine_image):
    # Initialize lists to store even and odd split data
    even = []
    odd = []
    
    # Split trials based on identifier
    for video, group in trials_df.groupby('identifier'):
        group = group.reset_index(drop=True)
        if len(group) > 1:
            
            odd_rows = group[group.index % 2 != 0]
            even_rows = group[group.index % 2 == 0]
            '''
            odd_rows = group[group.index.isin([0, 2,4])]
            even_rows = group[group.index.isin([1,3])]
            '''
            odd.append(odd_rows['betas'].mean())
            even.append(even_rows['betas'].mean())

    # Convert even and odd lists to DataFrames with a single column 'betas'
    even = pd.DataFrame({'betas': pd.Series(list(even))})
    odd = pd.DataFrame({'betas': pd.Series(list(odd))})

    # Use ROI_repr_all to get ROI representations from even and odd splits
    ROIeven_repr_df = even['betas'].apply(lambda x: ROI_repr_all(x, affine_image, subject_masks, roi_names))
    ROIodd_repr_df = odd['betas'].apply(lambda x: ROI_repr_all(x, affine_image, subject_masks, roi_names))

    # Initialize dictionary to store split-half RDM correlations for each ROI
    splithalf_rsa_r = {}
    for roi_name in roi_names:
        roi_size = len(np.where(subject_masks[roi_name[0]][roi_name[1]].get_fdata() == 1)[0]) #sometimes after reliabilitiy mask it becomes too small to do RSA
        if roi_size > 2:
            # Generate RDMs for the ROI in even and odd splits
            ROIevenbetas_rdm_vector = get_rdm(ROIeven_repr_df[roi_name[1]].to_list(), plot=False, method="pearsonr")
            ROIoddbetas_rdm_vector = get_rdm(ROIodd_repr_df[roi_name[1]].to_list(), plot=False,method="pearsonr")

            # Compute Spearman correlation between the RDMs as the split-half RSA noise ceiling
            r, p = scipy.stats.spearmanr(ROIevenbetas_rdm_vector, ROIoddbetas_rdm_vector)
            splithalf_rsa_r[roi_name[1]] = r  # Store the correlation value for the current ROI
        else:
            splithalf_rsa_r[roi_name[1]] = np.nan
        
    return splithalf_rsa_r
            
```
